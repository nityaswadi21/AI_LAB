```
from copy import deepcopy

# Goal state
GOAL = [[1, 2, 3],
        [4, 5, 6],
        [7, 8, 0]]

# Possible moves: Up, Down, Left, Right
MOVES = [(-1, 0), (1, 0), (0, -1), (0, 1)]


def print_state(state):
    for row in state:
        print(" ".join(str(x) if x != 0 else "_" for x in row))
    print()


def find_blank(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j


def get_neighbors(state):
    x, y = find_blank(state)

    for dx, dy in MOVES:
        nx, ny = x + dx, y + dy

        if 0 <= nx < 3 and 0 <= ny < 3:
            new_state = deepcopy(state)

            # Move blank tile
            new_state[x][y], new_state[nx][ny] = \
                new_state[nx][ny], new_state[x][y]

            yield new_state


def state_to_tuple(state):
    return tuple(tuple(row) for row in state)


def dfs(start):
    stack = [(start, [start])]
    visited = set()

    iteration = 0

    while stack:
        state, path = stack.pop()

        state_tuple = state_to_tuple(state)

        if state_tuple in visited:
            continue

        visited.add(state_tuple)
        iteration += 1

        print(f"DFS Iteration {iteration}:")
        print_state(state)

        # Check goal
        if state == GOAL:
            print("Goal state reached!")
            print("Final Goal State:")
            print_state(state)
            return path

        # Add neighboring states
        neighbors = list(get_neighbors(state))

        # Reverse so that the first move is explored first
        for neighbor in reversed(neighbors):
            if state_to_tuple(neighbor) not in visited:
                stack.append((neighbor, path + [neighbor]))

    return None


# Initial state
START = [[1, 2, 3],
         [4, 0, 6],
         [7, 5, 8]]

print("Initial State:")
print_state(START)

solution = dfs(START)

if solution:
    print("Solution Path:")
    for i, state in enumerate(solution):
        print(f"Step {i}:")
        print_state(state)
else:
    print("No solution found.")
```

```
from copy import deepcopy

# Goal state
GOAL = [[1, 2, 3],
        [4, 5, 6],
        [7, 8, 0]]

# Possible moves
MOVES = [(-1, 0), (1, 0), (0, -1), (0, 1)]


def print_state(state):
    for row in state:
        print(" ".join(str(x) if x != 0 else "_" for x in row))
    print()


def find_blank(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j


def get_neighbors(state):
    x, y = find_blank(state)

    for dx, dy in MOVES:
        nx, ny = x + dx, y + dy

        if 0 <= nx < 3 and 0 <= ny < 3:
            new_state = deepcopy(state)

            new_state[x][y], new_state[nx][ny] = \
                new_state[nx][ny], new_state[x][y]

            yield new_state


def state_to_tuple(state):
    return tuple(tuple(row) for row in state)


def depth_limited_search(state, depth, path, visited):
    print(f"Depth {len(path) - 1}:")
    print_state(state)

    # Goal test
    if state == GOAL:
        return path

    # Depth limit reached
    if depth == 0:
        return None

    state_tuple = state_to_tuple(state)
    visited.add(state_tuple)

    for neighbor in get_neighbors(state):
        neighbor_tuple = state_to_tuple(neighbor)

        # Avoid states already in the current path
        if neighbor_tuple not in visited:
            result = depth_limited_search(
                neighbor,
                depth - 1,
                path + [neighbor],
                visited
            )

            if result:
                return result

    visited.remove(state_tuple)

    return None


def ids(start):
    depth = 0

    while True:
        print("\n" + "=" * 40)
        print(f"IDS ITERATION - DEPTH LIMIT: {depth}")
        print("=" * 40)

        visited = set()

        result = depth_limited_search(
            start,
            depth,
            [start],
            visited
        )

        if result:
            print("\nGoal state reached!")
            print("\nFinal Goal State:")
            print_state(result[-1])

            return result

        depth += 1


# Initial state
START = [[1, 2, 3],
         [4, 0, 6],
         [7, 5, 8]]

print("Initial State:")
print_state(START)

solution = ids(START)

print("Solution Path:")
for i, state in enumerate(solution):
    print(f"Step {i}:")
    print_state(state)
```
