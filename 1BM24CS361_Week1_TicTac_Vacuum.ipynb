```
# Vacuum Cleaner - Goal Based Agent

def goal_based_agent(room, location):

    # Goal: Both rooms should be clean
    if room["A"] == "Clean" and room["B"] == "Clean":
        return "Goal Achieved"

    # If current room is dirty, clean it
    if room[location] == "Dirty":
        return "Suck"

    # If current room is clean, move to the other room
    if location == "A":
        return "Move Right"
    else:
        return "Move Left"


def vacuum_cleaner_goal_based():

    # Initial environment
    room = {
        "A": "Dirty",
        "B": "Dirty"
    }

    location = "A"

    print("Initial Environment:")
    print(room)
    print("Initial Location:", location)
    print()

    while True:

        action = goal_based_agent(room, location)

        print("Location:", location)
        print("Environment:", room)
        print("Action:", action)

        # Check goal
        if action == "Goal Achieved":
            break

        # Perform action
        if action == "Suck":
            room[location] = "Clean"

        elif action == "Move Right":
            location = "B"

        elif action == "Move Left":
            location = "A"

        print()

    print()
    print("GOAL ACHIEVED!")
    print("Final Environment:", room)


vacuum_cleaner_goal_based()
```

```
# Tic-Tac-Toe Game
# Both players are controlled by the program
# No standard AI algorithm is used

board = [" " for _ in range(9)]


def display_board():
    print()
    print(" {} | {} | {} ".format(board[0], board[1], board[2]))
    print("---+---+---")
    print(" {} | {} | {} ".format(board[3], board[4], board[5]))
    print("---+---+---")
    print(" {} | {} | {} ".format(board[6], board[7], board[8]))
    print()


def check_winner(player):
    winning_positions = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in winning_positions:
        if (board[a] == player and
            board[b] == player and
            board[c] == player):
            return True

    return False


def make_move(player):
    # Simple rule: choose the first available position
    for i in range(9):
        if board[i] == " ":
            board[i] = player
            return i

    return -1


def tic_tac_toe():
    print("TIC-TAC-TOE")
    print("Both players are controlled by the program.")

    current_player = "X"

    for turn in range(9):

        # Program makes the move
        position = make_move(current_player)

        print("\nPlayer", current_player, "chooses position", position + 1)

        display_board()

        # Check whether current player has won
        if check_winner(current_player):
            print("Player", current_player, "wins!")
            return

        # Change player
        if current_player == "X":
            current_player = "O"
        else:
            current_player = "X"

    print("Game Draw!")


tic_tac_toe()
```
