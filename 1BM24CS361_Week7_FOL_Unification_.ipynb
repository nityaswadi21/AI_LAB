```python

# Unification Algorithm in First Order Logic

def is_variable(x):
    return isinstance(x, str) and x.islower()


def apply_substitution(expr, subst):
    if isinstance(expr, tuple):
        return (expr[0],
                *[apply_substitution(arg, subst)
                  for arg in expr[1:]])

    while expr in subst:
        expr = subst[expr]

    return expr


def occurs_check(var, expr, subst):
    expr = apply_substitution(expr, subst)

    if var == expr:
        return True

    if isinstance(expr, tuple):
        return any(occurs_check(var, arg, subst)
                   for arg in expr[1:])

    return False


def unify(x, y, subst=None):

    if subst is None:
        subst = {}

    x = apply_substitution(x, subst)
    y = apply_substitution(y, subst)

    # If both expressions are identical
    if x == y:
        return subst

    # If x is a variable
    if is_variable(x):
        if occurs_check(x, y, subst):
            return None
        subst[x] = y
        return subst

    # If y is a variable
    if is_variable(y):
        if occurs_check(y, x, subst):
            return None
        subst[y] = x
        return subst

    # If both are compound expressions
    if isinstance(x, tuple) and isinstance(y, tuple):

        # Check predicate symbols
        if x[0] != y[0]:
            return None

        # Check number of arguments
        if len(x) != len(y):
            return None

        for a, b in zip(x[1:], y[1:]):
            subst = unify(a, b, subst)

            if subst is None:
                return None

        return subst

    return None


# Example expressions:
# Eats(x, Apple)
# Eats(Riya, y)

expr1 = ("Eats", "x", "Apple")
expr2 = ("Eats", "Riya", "y")

print("Expression 1:", expr1)
print("Expression 2:", expr2)

result = unify(expr1, expr2)

if result is not None:
    print("\nUnification Successful!")
    print("Substitution:", result)

    print("Unified Expression 1:",
          apply_substitution(expr1, result))

    print("Unified Expression 2:",
          apply_substitution(expr2, result))

else:
    print("\nUnification Failed!")
```
